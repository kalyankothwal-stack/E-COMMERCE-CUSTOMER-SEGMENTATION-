### Introduction

In today’s highly competitive e-commerce industry, understanding customer purchasing behavior is crucial for improving marketing strategies, increasing customer retention, and maximizing revenue. Customers differ significantly in how often they purchase, how recently they engage, and how much they spend.

This project focuses on segmenting e-commerce customers based on their purchasing behavior using machine learning techniques. By identifying distinct customer segments and predicting customer categories, businesses can design targeted marketing campaigns, optimize inventory management, and enhance overall customer satisfaction.

### Business Problem Statement

The e-commerce company currently lacks a structured approach to analyze customer purchasing behavior. Without proper segmentation, marketing efforts are often generic, leading to inefficient resource allocation and reduced customer engagement.

The objective of this project is to:
- Segment customers based on their purchasing patterns using unsupervised learning.
- Predict customer segments for new or existing customers.
- Provide actionable insights to improve marketing strategies and customer retention.


### Dataset Overview

The dataset contains transactional data from an online retail store, including customer purchases, product details, pricing, and purchase timestamps.

- Number of records: 541,909
- Number of features: 8
- Dataset Type: Transactional E-commerce Data
- Target Variable: Not available (Unsupervised Learning)

```python
def categorize_product(desc):
    if pd.isna(desc):
        return "Unknown"

    desc = desc.upper()

    for category, keywords in category_keywords.items():
        if any(keyword in desc for keyword in keywords):
            return category

    return "Other"
```

```python
df_clean['ProductCategory'] = df_clean['Description'].apply(categorize_product)
```

# 6.  CUSTOMER-LEVEL FEATURE ENGINEERING

```python
df_rfm = df_clean.copy()
```

```python
df_rfm
```

```python
# RFM Analysis on all data 
reference_date = df_rfm['InvoiceDate'].max() + timedelta(days=1)
reference_date
```

```python
# Calculate RFM metrics per customer
rfm_data = df_rfm.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (reference_date - x.max()).days,  # Recency
    'InvoiceNo': 'nunique',  # Frequency
    'TotalAmount': ['sum','mean']  # Monetary
}).reset_index()

rfm_data.columns = ['CustomerID', 'Recency', 'Frequency', 'MonetaryTotal', 'MonetaryAvg']
rfm_data
```

```python
# Calculate return-related features per customer
return_features = df_clean[df_clean['IsReturn']].groupby('CustomerID').agg({
    'InvoiceNo': 'nunique',
    'Quantity': lambda x: abs(x).sum(),
    'TotalAmount': lambda x: abs(x).sum()
}).reset_index()
return_features.columns = ['CustomerID', 'ReturnFrequency', 'ReturnQuantity', 'ReturnAmount']
return_features.head()
```

```python
# Merge with RFM data
rfm_data = rfm_data.merge(return_features, on='CustomerID', how='left')
rfm_data
```

```python
# Fill NaN values with 0 (customers who never returned)
rfm_data[['ReturnFrequency', 'ReturnQuantity', 'ReturnAmount']] = rfm_data[['ReturnFrequency', 'ReturnQuantity', 'ReturnAmount']].fillna(0)
```

```python
# Calculate return rate
rfm_data['ReturnRate'] = rfm_data['ReturnAmount'] / (rfm_data['MonetaryTotal'])
rfm_data['ReturnRate'] = rfm_data['ReturnRate'].fillna(0)
```

```python
print(f"✓ Return features added: ReturnFrequency, ReturnQuantity, ReturnAmount, ReturnRate")
print(f"✓ Customers with returns: {(rfm_data['ReturnFrequency'] > 0).sum()}")
```

```python
# Monthly purchase amount
rfm_data['MonthlyPurchaseAmount'] = rfm_data['MonetaryTotal'] / 12

print(f"✓ RFM features created for {rfm_data.shape[0]} customers")
rfm_data.head()
```

```python
# Category Preferences per Customer
category_customer = df_clean.groupby(['CustomerID', 'ProductCategory'])['TotalAmount'].sum().reset_index()
category_customer
```

```python
# Get top and bottom category for each customer
top_category = category_customer.loc[category_customer.groupby('CustomerID')['TotalAmount'].idxmax()]
top_category = top_category[['CustomerID', 'ProductCategory', 'TotalAmount']].rename(
    columns={'ProductCategory': 'TopCategory', 'TotalAmount': 'TopCategorySpend'})
```

```python
top_category
```

```python
# Merge category preferences
rfm_data = rfm_data.merge(top_category, on='CustomerID', how='left')
# rfm_data = rfm_data.merge(bottom_category, on='CustomerID', how='left')

print(f"✓ Category preferences added")
```

```python
rfm_data
```

```python
# Country information
customer_country = df_clean.groupby('CustomerID')['Country'].agg(lambda x: x.mode()[0]).reset_index()
rfm_data = rfm_data.merge(customer_country, on='CustomerID', how='left')
```

```python
# Total quantity purchased
total_quantity = df_clean.groupby('CustomerID')['Quantity'].sum().reset_index()
total_quantity.columns = ['CustomerID', 'TotalQuantity']
rfm_data = rfm_data.merge(total_quantity, on='CustomerID', how='left')
```

```python
# Purchase frequency (transactions per month)
rfm_data['PurchaseFrequencyPerMonth'] = rfm_data['Frequency'] / 12
```

```python
# Customer lifetime metrics (using all data)
lifetime_metrics = df_clean.groupby('CustomerID').agg({
    'InvoiceDate': ['min', 'max'],
    'TotalAmount': 'sum',
    'InvoiceNo': 'nunique'
}).reset_index()
lifetime_metrics.columns = ['CustomerID', 'FirstPurchase', 'LastPurchase', 
                            'LifetimeValue', 'LifetimeFrequency']
lifetime_metrics['CustomerAge'] = (lifetime_metrics['LastPurchase'] - 
                                  lifetime_metrics['FirstPurchase']).dt.days

rfm_data = rfm_data.merge(lifetime_metrics[['CustomerID', 'CustomerAge', 'LifetimeValue', 'LifetimeFrequency']], 
                         on='CustomerID', how='left')
```

```python
# Loyal customer indicator (customers with > 10 transactions and > 90 days active)
rfm_data['IsLoyal'] = ((rfm_data['LifetimeFrequency'] > 10) & 
                       (rfm_data['CustomerAge'] > 90)).astype(int)
```

```python
# Frequent buyer indicator (> 1 purchase per month)
rfm_data['IsFrequentBuyer'] = (rfm_data['PurchaseFrequencyPerMonth'] > 1).astype(int)

print(f"✓ Total features created: {rfm_data.shape[1]}")
print(f"✓ Loyal customers identified: {rfm_data['IsLoyal'].sum()}")
print(f"✓ Frequent buyers identified: {rfm_data['IsFrequentBuyer'].sum()}")

rfm_data
```

```python
# Check how many inf values exist
inf_count = np.isinf(rfm_data['ReturnRate']).sum()
print(f"Infinite ReturnRate values before fix: {inf_count}")

# Replace inf and -inf with 0, then handle NaN
rfm_data['ReturnRate'] = (rfm_data['ReturnRate'].replace([np.inf, -np.inf], 0).fillna(0))

# Verify
print("Infinite ReturnRate values after fix:",
      np.isinf(rfm_data['ReturnRate']).sum())
```

# 7. OUTLIER DETECTION AND REMOVAL

```python
# Define columns to check for outliers
outlier_columns = ['Recency', 'Frequency', 'MonetaryTotal', 'MonetaryAvg','PurchaseFrequencyPerMonth',
                   'CustomerAge', 'TotalQuantity', 'IsLoyal', 'ReturnRate' ]

print(f"Original dataset: {rfm_data.shape[0]} customers")

n = len(outlier_columns)   # number of columns to plot
cols = 3                   # 3 plots per row
rows = math.ceil(n / cols) # enough rows to fit all

fig, axes = plt.subplots(rows, cols, figsize=(5*cols, 4*rows))
axes = axes.flatten()

for idx, col in enumerate(outlier_columns):
    axes[idx].boxplot(rfm_data[col].dropna())
    axes[idx].set_title(f'{col}', fontsize=10, fontweight='bold')
    axes[idx].set_ylabel('Value')
    axes[idx].grid(True, alpha=0.3)

# Hide unused axes
for j in range(len(outlier_columns), len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.savefig('outliers_boxplot.png', dpi=300, bbox_inches='tight')
plt.show()
print("✓ Outlier boxplots saved")

# Remove outliers using percentile method (removes extreme 1% on each end)
def remove_outliers_percentile(df, columns, lower=1, upper=99):
    """Remove top and bottom percentiles"""
    df_clean = df.copy()
    for col in columns:
        before = len(df_clean)
        lower_bound = df[col].quantile(lower/100)
        upper_bound = df[col].quantile(upper/100)
        df_clean = df_clean[(df_clean[col] >= lower_bound) & (df_clean[col] <= upper_bound)]
        after = len(df_clean)
        print(f"  {col}: Removed {before - after} outliers")
    return df_clean

rfm_clean = remove_outliers_percentile(rfm_data, outlier_columns, lower=1, upper=99)

print(f"\n✓ After outlier removal: {rfm_clean.shape[0]} customers")
print(f"✓ Total removed: {rfm_data.shape[0] - rfm_clean.shape[0]} ({(rfm_data.shape[0] - rfm_clean.shape[0])/rfm_data.shape[0]*100:.1f}%)")

# Update rfm_data with clean data
rfm_data = rfm_clean.reset_index(drop=True)

print("✓ Outliers handled successfully")
```

# 9. CLUSTERING - MODEL SELECTION

```python
# Test different number of clusters for K-Means
inertias = []
silhouette_scores = []
davies_bouldin_scores = []
calinski_harabasz_scores = []

K_range = range(2, 11)

print("\n[Evaluating K-Means for different K values]")
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    
    inertias.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(X_scaled, labels))
    davies_bouldin_scores.append(davies_bouldin_score(X_scaled, labels))
    calinski_harabasz_scores.append(calinski_harabasz_score(X_scaled, labels))
    
    print(f"K={k}: Silhouette={silhouette_scores[-1]:.3f}, "
          f"Davies-Bouldin={davies_bouldin_scores[-1]:.3f}, "
          f"Calinski-Harabasz={calinski_harabasz_scores[-1]:.1f}")
```

```python
# Plot evaluation metrics
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

axes[0, 0].plot(K_range, inertias, 'bo-')
axes[0, 0].set_xlabel('Number of Clusters')
axes[0, 0].set_ylabel('Inertia')
axes[0, 0].set_title('Elbow Method')
axes[0, 0].grid(True)

axes[0, 1].plot(K_range, silhouette_scores, 'ro-')
axes[0, 1].set_xlabel('Number of Clusters')
axes[0, 1].set_ylabel('Silhouette Score')
axes[0, 1].set_title('Silhouette Score (Higher is Better)')
axes[0, 1].grid(True)

axes[1, 0].plot(K_range, davies_bouldin_scores, 'go-')
axes[1, 0].set_xlabel('Number of Clusters')
axes[1, 0].set_ylabel('Davies-Bouldin Score')
axes[1, 0].set_title('Davies-Bouldin Score (Lower is Better)')
axes[1, 0].grid(True)

axes[1, 1].plot(K_range, calinski_harabasz_scores, 'mo-')
axes[1, 1].set_xlabel('Number of Clusters')
axes[1, 1].set_ylabel('Calinski-Harabasz Score')
axes[1, 1].set_title('Calinski-Harabasz Score (Higher is Better)')
axes[1, 1].grid(True)

plt.tight_layout()
plt.savefig('clustering_evaluation.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Select optimal K (you can adjust based on the plots)
optimal_k = 4  # Based on silhouette score and business interpretation
print(f"\n✓ Optimal number of clusters selected: {optimal_k}")
```

# 10.  FINAL CLUSTERING MODELS

```python
# 1. K-Means Clustering
kmeans_final = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
kmeans_labels = kmeans_final.fit_predict(X_scaled)
clustering_features['KMeans_Cluster'] = kmeans_labels

print(f"✓ K-Means clustering completed")
print(f"  Silhouette Score: {silhouette_score(X_scaled, kmeans_labels):.3f}")
```

```python
# 2. Hierarchical Clustering
hierarchical = AgglomerativeClustering(n_clusters=optimal_k)
hierarchical_labels = hierarchical.fit_predict(X_scaled)
clustering_features['Hierarchical_Cluster'] = hierarchical_labels

print(f"✓ Hierarchical clustering completed")
print(f"  Silhouette Score: {silhouette_score(X_scaled, hierarchical_labels):.3f}")
```

```python
# 3. DBSCAN (for comparison)
dbscan = DBSCAN(eps=2, min_samples=5)
dbscan_labels = dbscan.fit_predict(X_scaled)
clustering_features['DBSCAN_Cluster'] = dbscan_labels

n_clusters_dbscan = len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0)
print(f"✓ DBSCAN clustering completed")
print(f"  Number of clusters: {n_clusters_dbscan}")
print(f"  Noise points: {list(dbscan_labels).count(-1)}")
```

```python
# Use K-Means as final model
final_labels = kmeans_labels
clustering_features['FinalCluster'] = final_labels
```

# 13. BUSINESS RECOMMENDATIONS

```python
# FINAL CLUSTER NAMING & BUSINESS INSIGHTS (K = 4)

cluster_insights = {}

# ---------- Step 1: Cluster-level summary ----------
cluster_summary = clustering_features.groupby('FinalCluster').agg({
    'Recency': 'mean',
    'Frequency': 'mean',
    'MonetaryTotal': 'mean',
    'IsLoyal': 'mean'
}).reset_index()

# ---------- Step 2: Rank clusters (for unique naming) ----------
cluster_summary['MonetaryRank'] = cluster_summary['MonetaryTotal'].rank(ascending=False)
cluster_summary['FrequencyRank'] = cluster_summary['Frequency'].rank(ascending=False)
cluster_summary['RecencyRank'] = cluster_summary['Recency'].rank(ascending=True)

# ---------- Step 3: Assign unique cluster names ----------
cluster_names = {}

for _, row in cluster_summary.iterrows():
    c = int(row['FinalCluster'])

    if row['MonetaryRank'] == 1 and row['FrequencyRank'] == 1:
        name = "Champions"
        recommendation = "Reward with VIP offers, exclusive deals, and early access."

    elif row['FrequencyRank'] <= 2:
        name = "Loyal Customers"
        recommendation = "Upsell premium products and strengthen loyalty programs."

    elif row['RecencyRank'] >= 3:
        name = "At-Risk Customers"
        recommendation = "Run win-back campaigns with discounts and re-engagement emails."

    else:
        name = "Occasional Buyers"
        recommendation = "Send personalized offers and reminders to increase frequency."

    cluster_names[c] = {
        "Name": name,
        "Recommendation": recommendation
    }

# ---------- Step 4: Build final insights ----------
for cluster in range(optimal_k):
    cluster_data = clustering_features[
        clustering_features['FinalCluster'] == cluster
    ]

    avg_recency = cluster_data['Recency'].mean()
    avg_frequency = cluster_data['Frequency'].mean()
    avg_monetary = cluster_data['MonetaryTotal'].mean()
    loyal_pct = cluster_data['IsLoyal'].mean() * 100

    cluster_insights[cluster] = {
        'Name': cluster_names[cluster]['Name'],
        'Size': len(cluster_data),
        'Percentage': f"{len(cluster_data) / len(clustering_features) * 100:.1f}%",
        'Avg_Recency': f"{avg_recency:.0f} days",
        'Avg_Frequency': f"{avg_frequency:.1f} transactions",
        'Avg_Monetary': f"£{avg_monetary:.2f}",
        'Loyal_Percentage': f"{loyal_pct:.1f}%",
        'Top_Category': (
            cluster_data['TopCategory'].mode()[0]
            if not cluster_data['TopCategory'].mode().empty
            else 'N/A'
        ),
        'Recommendation': cluster_names[cluster]['Recommendation']
    }

# ---------- Step 5: Attach names back to dataframe ----------
clustering_features['ClusterName'] = clustering_features['FinalCluster'].map(
    {k: v['Name'] for k, v in cluster_names.items()}
)

# ---------- Step 6: Display insights ----------
for c, info in cluster_insights.items():
    print(f"\nCLUSTER {c}: {info['Name']}")
    for k, v in info.items():
        if k != 'Name':
            print(f"  {k}: {v}")
```

```python
# Save insights to CSV
insights_df = pd.DataFrame(cluster_insights).T
insights_df.to_csv('cluster_insights.csv')
print("\n✓ Cluster insights saved to 'cluster_insights.csv'")
```

# 16.  DEMONSTRATION - PREDICTING NEW CUSTOMER CLUSTER

```python
import numpy as np
import joblib

# Define feature columns
feature_cols = [
    'Recency', 'Frequency', 'MonetaryTotal', 'MonetaryAvg',
    'PurchaseFrequencyPerMonth', 'CustomerAge', 'IsLoyal',
    'TopCategory_Encoded', 'TotalQuantity', 'ReturnRate'
]

def predict_customer_cluster(customer_data):
    """
    Predict cluster for a new customer
    
    Parameters:
    customer_data: dict with customer features
    
    Returns:
    Predicted cluster, cluster name, and recommendation
    """
    # Load models and encoders
    kmeans = joblib.load('kmeans_model.pkl')
    rf_clf = joblib.load('cluster_classifier.pkl')
    scaler = joblib.load('robust_scaler.pkl')
    le_top = joblib.load('label_encoder_top_category.pkl')
    
    # Prepare features
    features = []
    for col in feature_cols:
        if col == 'TopCategory_Encoded':
            category = customer_data.get('TopCategory', 'Unknown')
            if category in le_top.classes_:
                val = le_top.transform([category])[0]
            else:
                val = -1  # unseen category handling
        else:
            # Numerical features
            val = customer_data.get(col, 0)
        features.append(val)
    
    # Scale features
    features_array = np.array(features).reshape(1, -1)
    features_scaled = scaler.transform(features_array)
    
    # Predict cluster
    cluster = rf_clf.predict(features_scaled)[0]
    cluster_name = cluster_insights[cluster]['Name']
    recommendation = cluster_insights[cluster]['Recommendation']
    
    return cluster, cluster_name, recommendation


# Example prediction
print("\n[Example: New Customer Prediction]")
new_customer = {
    'Recency': 30,
    'Frequency': 15,
    'MonetaryTotal': 2500,
    'MonetaryAvg': 166.67,
    'PurchaseFrequencyPerMonth': 2.5,
    'CustomerAge': 180,
    'IsLoyal': 1,
    'TopCategory': 'Kitchen & Dining',
    'TotalQuantity': 10,
    'ReturnRate': 0.05
}

cluster_pred, cluster_name, recommendation = predict_customer_cluster(new_customer)

print(f"\nCustomer Profile:")
for key, value in new_customer.items():
    print(f"  {key}: {value}")

print(f"\n✓ Predicted Cluster: {cluster_pred}")
print(f"✓ Cluster Name: {cluster_name}")
print(f"✓ Recommendation: {recommendation}")

print("\n" + "="*80)
print("END OF ANALYSIS")
print("="*80)
```
