4. Feature Engineering

6. Customer Segmentation
7. Prediction & Evaluation
8. Business Insights


### Step 1
This step performs part of the data science pipeline. The code below is **unchanged from the original notebook**, but is now explained clearly for evaluation and viva purposes.

```
"""
E-COMMERCE CUSTOMER SEGMENTATION AND PREDICTION
Complete Machine Learning Pipeline
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

# Machine Learning
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.decomposition import PCA
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import joblib

plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

print("="*80)
print("E-COMMERCE CUSTOMER SEGMENTATION PROJECT")
print("="*80)

# ============================================================================
# SECTION 1: DATA LOADING AND PRODUCT CATEGORIZATION
# ============================================================================

print("\n[STEP 1] LOADING AND EXPLORING DATA")
print("-"*80)

# Load dataset
df = pd.read_csv('e-commerce data.csv', encoding='ISO-8859-1')
print(f"✓ Dataset loaded: {df.shape[0]} rows, {df.shape[1]} columns")
print(f"\nColumns: {df.columns.tolist()}")

# Basic Info
print("\n[Initial Data Overview]")
print(df.head())
print("\n[Data Types]")
print(df.dtypes)
print("\n[Missing Values]")
print(df.isnull().sum())

# ============================================================================
# PRODUCT CATEGORIZATION MAPPING
# ============================================================================

print("\n[STEP 2] PRODUCT CATEGORIZATION")
print("-"*80)

# Comprehensive product categorization based on provided product list
def categorize_product(description):
    """Categorize products into main categories"""
    if pd.isna(description):
        return 'Unknown'
    
    desc = str(description).upper()
    
    # Home & Kitchen
    if any(word in desc for word in ['KITCHEN', 'COOK', 'DINNER', 'PLATE', 'BOWL', 'MUG', 'CUP', 
                                      'SPOON', 'FORK', 'KNIFE', 'CUTLERY', 'PAN', 'POT', 'KETTLE',
                                      'OVEN', 'MICROWAVE', 'JAR', 'STORAGE', 'CANISTER', 'BREAD BIN',
                                      'TEA', 'COFFEE', 'MILK JUG', 'SUGAR', 'SALT', 'PEPPER']):
        return 'Kitchen & Dining'
    
    # Home Decor
    elif any(word in desc for word in ['CANDLE', 'T-LIGHT', 'LANTERN', 'LAMP', 'LIGHT', 'FRAME',
                                        'MIRROR', 'CLOCK', 'WALL', 'DECORATION', 'WREATH', 'GARLAND',
                                        'CUSHION', 'PILLOW', 'CURTAIN', 'DOOR', 'WINDOW', 'CABINET',
                                        'SHELF', 'DRAWER', 'HOOK', 'HANGER', 'HOLDER']):
        return 'Home Decor'
    
    # Bags & Accessories
    elif any(word in desc for word in ['BAG', 'HANDBAG', 'PURSE', 'RUCKSACK', 'BACKPACK', 'TOTE',
                                        'LUGGAGE', 'SUITCASE', 'WALLET', 'POUCH']):
        return 'Bags & Accessories'
    
    # Fashion & Jewelry
    elif any(word in desc for word in ['NECKLACE', 'BRACELET', 'EARRING', 'RING', 'JEWEL', 'BROOCH',
                                        'SCARF', 'HAT', 'UMBRELLA', 'GLOVE', 'SOCK', 'SHOE', 'SLIPPER',
                                        'APRON', 'BANGLE', 'CHARM', 'PENDANT']):
        return 'Fashion & Jewelry'
    
    # Stationery & Office
    elif any(word in desc for word in ['PEN', 'PENCIL', 'NOTEBOOK', 'CARD', 'PAPER', 'WRAP', 'GIFT',
                                        'STICKER', 'STAMP', 'ENVELOPE', 'TAPE', 'GLUE', 'RULER',
                                        'ERASER', 'CRAYON', 'PAINT', 'CHALK']):
        return 'Stationery & Gifts'
    
    # Toys & Games
    elif any(word in desc for word in ['TOY', 'GAME', 'DOLL', 'PUZZLE', 'BALL', 'TEDDY', 'SOFT TOY',
                                        'PLAY', 'CHILDREN', 'KIDS', 'BABY', 'SPACEBOY', 'DOLLY GIRL',
                                        'CIRCUS', 'ROBOT']):
        return 'Toys & Games'
    
    # Garden & Outdoor
    elif any(word in desc for word in ['GARDEN', 'PLANT', 'FLOWER', 'SEED', 'WATERING', 'BIRD',
                                        'OUTDOOR', 'PARASOL', 'PICNIC', 'BASKET', 'HAMMOCK']):
        return 'Garden & Outdoor'
    
    # Bathroom & Personal Care
    elif any(word in desc for word in ['BATHROOM', 'SOAP', 'TOWEL', 'SHOWER', 'BATH', 'COSMETIC',
                                        'PERFUME', 'TOILETTE', 'HAIR', 'COMB', 'BRUSH']):
        return 'Bathroom & Personal Care'
    
    # Party & Celebration
    elif any(word in desc for word in ['PARTY', 'BIRTHDAY', 'CAKE', 'BALLOON', 'BUNTING', 'CELEBRATION',
                                        'WEDDING', 'CHRISTMAS', 'EASTER', 'HALLOWEEN']):
        return 'Party & Celebration'
    
    # Textiles & Fabrics
    elif any(word in desc for word in ['FABRIC', 'CLOTH', 'TEXTILE', 'LINEN', 'COTTON', 'RIBBON',
                                        'SEWING', 'KNIT', 'CROCHET', 'EMBROIDERY', 'FELT']):
        return 'Textiles & Crafts'
    
    else:
        return 'General Merchandise'

# Apply categorization
df['Category'] = df['Description'].apply(categorize_product)
print(f"✓ Products categorized into {df['Category'].nunique()} categories")
print("\n[Category Distribution]")
print(df['Category'].value_counts())

# ============================================================================
# SECTION 2: DATA CLEANING AND PREPROCESSING
# ============================================================================

print("\n[STEP 3] DATA CLEANING")
print("-"*80)

# Store original shape
original_shape = df.shape[0]

# Convert InvoiceDate to datetime
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], format= 'mixed')

# Remove records with missing CustomerID
df = df[df['CustomerID'].notna()]
print(f"✓ Removed {original_shape - df.shape[0]} rows with missing CustomerID")

# Remove negative quantities (returns will be handled separately)
df_clean = df[df['Quantity'] > 0].copy()

# Remove negative prices
df_clean = df_clean[df_clean['UnitPrice'] > 0]

# Calculate Total Amount
df_clean['TotalAmount'] = df_clean['Quantity'] * df_clean['UnitPrice']

# Identify returns (InvoiceNo starting with 'C')
df['IsReturn'] = df['InvoiceNo'].astype(str).str.startswith('C')
total_returns = df[df['IsReturn']]['Quantity'].abs().sum()

print(f"✓ Total Quantity Sold: {df_clean['Quantity'].sum():,.0f}")
print(f"✓ Total Quantity Returned: {total_returns:,.0f}")
print(f"✓ Clean dataset: {df_clean.shape[0]} rows")

# Get last 6 months of data for RFM analysis
max_date = df_clean['InvoiceDate'].max()
six_months_ago = max_date - timedelta(days=180)
df_rfm = df_clean[df_clean['InvoiceDate'] >= six_months_ago].copy()

print(f"✓ RFM Analysis period: {six_months_ago.date()} to {max_date.date()}")
print(f"✓ RFM dataset: {df_rfm.shape[0]} rows")

# ============================================================================
# SECTION 3: EXPLORATORY DATA ANALYSIS
# ============================================================================

print("\n[STEP 4] EXPLORATORY DATA ANALYSIS")
print("-"*80)

# Revenue Analysis
print("\n[Revenue Insights]")
total_revenue = df_clean['TotalAmount'].sum()
print(f"Total Revenue: £{total_revenue:,.2f}")

# Top 10 Products by Revenue
top_products = df_clean.groupby('Description').agg({
    'TotalAmount': 'sum',
    'Quantity': 'sum'
}).sort_values('TotalAmount', ascending=False).head(10)
print("\n[Top 10 Products by Revenue]")
print(top_products)

# Highest Priced Products
highest_priced = df_clean.nlargest(10, 'UnitPrice')[['Description', 'UnitPrice', 'Category']]
print("\n[Highest Priced Products]")
print(highest_priced)

# Country Analysis
country_revenue = df_clean.groupby('Country')['TotalAmount'].sum().sort_values(ascending=False)
print("\n[Top 10 Countries by Revenue]")
print(country_revenue.head(10))

# Category Analysis
category_revenue = df_clean.groupby('Category')['TotalAmount'].sum().sort_values(ascending=False)
print("\n[Revenue by Category]")
print(category_revenue)

# Visualizations
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Revenue Over Time
monthly_revenue = df_clean.groupby(df_clean['InvoiceDate'].dt.to_period('M'))['TotalAmount'].sum()
monthly_revenue.index = monthly_revenue.index.to_timestamp()
axes[0, 0].plot(monthly_revenue.index, monthly_revenue.values, marker='o', linewidth=2, markersize=6)
axes[0, 0].set_title('Monthly Revenue Trend', fontsize=14, fontweight='bold')
axes[0, 0].set_xlabel('Month')
axes[0, 0].set_ylabel('Revenue (£)')
axes[0, 0].grid(True, alpha=0.3)
axes[0, 0].tick_params(axis='x', rotation=45)

# 2. Top 10 Products
top_10_products = df_clean.groupby('Description')['TotalAmount'].sum().nlargest(10)
axes[0, 1].barh(range(len(top_10_products)), top_10_products.values)
axes[0, 1].set_yticks(range(len(top_10_products)))
axes[0, 1].set_yticklabels([desc[:30] + '...' if len(desc) > 30 else desc 
                             for desc in top_10_products.index], fontsize=8)
axes[0, 1].set_title('Top 10 Products by Revenue', fontsize=14, fontweight='bold')
axes[0, 1].set_xlabel('Revenue (£)')

# 3. Category Distribution
category_revenue.plot(kind='bar', ax=axes[1, 0], color='skyblue', edgecolor='black')
axes[1, 0].set_title('Revenue by Category', fontsize=14, fontweight='bold')
axes[1, 0].set_xlabel('Category')
axes[1, 0].set_ylabel('Revenue (£)')
axes[1, 0].tick_params(axis='x', rotation=45)

# 4. Month-over-Month Growth
monthly_growth = monthly_revenue.pct_change() * 100
axes[1, 1].plot(monthly_growth.index, monthly_growth.values, marker='o', 
                linewidth=2, markersize=6, color='green')
axes[1, 1].axhline(y=0, color='r', linestyle='--', alpha=0.5)
axes[1, 1].set_title('Month-over-Month Revenue Growth (%)', fontsize=14, fontweight='bold')
axes[1, 1].set_xlabel('Month')
axes[1, 1].set_ylabel('Growth Rate (%)')
axes[1, 1].grid(True, alpha=0.3)
axes[1, 1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.savefig('eda_visualizations.png', dpi=300, bbox_inches='tight')
plt.show()
print("✓ EDA visualizations saved")

# ============================================================================
# SECTION 4: CUSTOMER-LEVEL FEATURE ENGINEERING
# ============================================================================

print("\n[STEP 5] CUSTOMER FEATURE ENGINEERING")
print("-"*80)

# RFM Analysis on last 6 months
reference_date = df_rfm['InvoiceDate'].max() + timedelta(days=1)

# Calculate RFM metrics per customer
rfm_data = df_rfm.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (reference_date - x.max()).days,  # Recency
    'InvoiceNo': 'nunique',  # Frequency
    'TotalAmount': ['sum', 'mean', 'min', 'max']  # Monetary
}).reset_index()

rfm_data.columns = ['CustomerID', 'Recency', 'Frequency', 'MonetaryTotal', 
                    'MonetaryAvg', 'MonetaryMin', 'MonetaryMax']

# Monthly purchase amount (avg per month over 6 months)
rfm_data['MonthlyPurchaseAmount'] = rfm_data['MonetaryTotal'] / 6

print(f"✓ RFM features created for {rfm_data.shape[0]} customers")

# Category Preferences per Customer
category_customer = df_clean.groupby(['CustomerID', 'Category'])['TotalAmount'].sum().reset_index()

# Get top and bottom category for each customer
top_category = category_customer.loc[category_customer.groupby('CustomerID')['TotalAmount'].idxmax()]
top_category = top_category[['CustomerID', 'Category', 'TotalAmount']].rename(
    columns={'Category': 'TopCategory', 'TotalAmount': 'TopCategorySpend'})

bottom_category = category_customer.loc[category_customer.groupby('CustomerID')['TotalAmount'].idxmin()]
bottom_category = bottom_category[['CustomerID', 'Category']].rename(
    columns={'Category': 'BottomCategory'})

# Merge category preferences
rfm_data = rfm_data.merge(top_category, on='CustomerID', how='left')
rfm_data = rfm_data.merge(bottom_category, on='CustomerID', how='left')

print(f"✓ Category preferences added")

# Country information
customer_country = df_clean.groupby('CustomerID')['Country'].agg(lambda x: x.mode()[0]).reset_index()
rfm_data = rfm_data.merge(customer_country, on='CustomerID', how='left')

# Total quantity purchased
total_quantity = df_clean.groupby('CustomerID')['Quantity'].sum().reset_index()
total_quantity.columns = ['CustomerID', 'TotalQuantity']
rfm_data = rfm_data.merge(total_quantity, on='CustomerID', how='left')

# Purchase frequency (transactions per month)
rfm_data['PurchaseFrequencyPerMonth'] = rfm_data['Frequency'] / 6

# Customer lifetime metrics (using all data)
lifetime_metrics = df_clean.groupby('CustomerID').agg({
    'InvoiceDate': ['min', 'max'],
    'TotalAmount': 'sum',
    'InvoiceNo': 'nunique'
}).reset_index()
lifetime_metrics.columns = ['CustomerID', 'FirstPurchase', 'LastPurchase', 
                            'LifetimeValue', 'LifetimeFrequency']
lifetime_metrics['CustomerAge'] = (lifetime_metrics['LastPurchase'] - 
                                  lifetime_metrics['FirstPurchase']).dt.days

rfm_data = rfm_data.merge(lifetime_metrics[['CustomerID', 'CustomerAge', 'LifetimeValue', 'LifetimeFrequency']], 
                         on='CustomerID', how='left')

# Loyal customer indicator (customers with > 10 transactions and > 90 days active)
rfm_data['IsLoyal'] = ((rfm_data['LifetimeFrequency'] > 10) & 
                       (rfm_data['CustomerAge'] > 90)).astype(int)

# Frequent buyer indicator (> 1 purchase per month)
rfm_data['IsFrequentBuyer'] = (rfm_data['PurchaseFrequencyPerMonth'] > 1).astype(int)

print(f"✓ Total features created: {rfm_data.shape[1]}")
print(f"✓ Loyal customers identified: {rfm_data['IsLoyal'].sum()}")
print(f"✓ Frequent buyers identified: {rfm_data['IsFrequentBuyer'].sum()}")

print("\n[Feature Summary]")
print(rfm_data.head())
print("\n[Feature Statistics]")
print(rfm_data.describe())

# Save intermediate features
rfm_data.to_csv('customer_features.csv', index=False)
print("✓ Customer features saved to 'customer_features.csv'")

# ============================================================================
# SECTION 5: DATA PREPARATION FOR CLUSTERING
# ============================================================================

print("\n[STEP 6] PREPARING DATA FOR CLUSTERING")
print("-"*80)

# Select features for clustering
clustering_features = rfm_data.copy()

# Encode categorical features
le_top_cat = LabelEncoder()
le_bottom_cat = LabelEncoder()
le_country = LabelEncoder()

clustering_features['TopCategory_Encoded'] = le_top_cat.fit_transform(
    clustering_features['TopCategory'].fillna('Unknown'))
clustering_features['BottomCategory_Encoded'] = le_bottom_cat.fit_transform(
    clustering_features['BottomCategory'].fillna('Unknown'))
clustering_features['Country_Encoded'] = le_country.fit_transform(
    clustering_features['Country'].fillna('Unknown'))

# Save encoders
joblib.dump(le_top_cat, 'label_encoder_top_category.pkl')
joblib.dump(le_bottom_cat, 'label_encoder_bottom_category.pkl')
joblib.dump(le_country, 'label_encoder_country.pkl')

# Select numerical features for clustering
feature_cols = ['Recency', 'Frequency', 'MonetaryTotal', 'MonetaryAvg', 'MonetaryMin', 
                'MonetaryMax', 'MonthlyPurchaseAmount', 'TopCategorySpend', 'TotalQuantity',
                'PurchaseFrequencyPerMonth', 'CustomerAge', 'LifetimeValue', 
                'LifetimeFrequency', 'IsLoyal', 'IsFrequentBuyer',
                'TopCategory_Encoded', 'BottomCategory_Encoded', 'Country_Encoded']

X = clustering_features[feature_cols].fillna(0)

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Save scaler
joblib.dump(scaler, 'standard_scaler.pkl')

print(f"✓ Features scaled: {X_scaled.shape}")
print(f"✓ Feature columns: {feature_cols}")

# ============================================================================
# SECTION 6: CLUSTERING - MODEL SELECTION
# ============================================================================

print("\n[STEP 7] CLUSTERING MODEL SELECTION")
print("-"*80)

# Test different number of clusters for K-Means
inertias = []
silhouette_scores = []
davies_bouldin_scores = []
calinski_harabasz_scores = []

K_range = range(2, 11)

print("\n[Evaluating K-Means for different K values]")
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    
    inertias.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(X_scaled, labels))
    davies_bouldin_scores.append(davies_bouldin_score(X_scaled, labels))
    calinski_harabasz_scores.append(calinski_harabasz_score(X_scaled, labels))
    
    print(f"K={k}: Silhouette={silhouette_scores[-1]:.3f}, "
          f"Davies-Bouldin={davies_bouldin_scores[-1]:.3f}, "
          f"Calinski-Harabasz={calinski_harabasz_scores[-1]:.1f}")

# Plot evaluation metrics
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

axes[0, 0].plot(K_range, inertias, 'bo-')
axes[0, 0].set_xlabel('Number of Clusters')
axes[0, 0].set_ylabel('Inertia')
axes[0, 0].set_title('Elbow Method')
axes[0, 0].grid(True)

axes[0, 1].plot(K_range, silhouette_scores, 'ro-')
axes[0, 1].set_xlabel('Number of Clusters')
axes[0, 1].set_ylabel('Silhouette Score')
axes[0, 1].set_title('Silhouette Score (Higher is Better)')
axes[0, 1].grid(True)

axes[1, 0].plot(K_range, davies_bouldin_scores, 'go-')
axes[1, 0].set_xlabel('Number of Clusters')
axes[1, 0].set_ylabel('Davies-Bouldin Score')
axes[1, 0].set_title('Davies-Bouldin Score (Lower is Better)')
axes[1, 0].grid(True)

axes[1, 1].plot(K_range, calinski_harabasz_scores, 'mo-')
axes[1, 1].set_xlabel('Number of Clusters')
axes[1, 1].set_ylabel('Calinski-Harabasz Score')
axes[1, 1].set_title('Calinski-Harabasz Score (Higher is Better)')
axes[1, 1].grid(True)

plt.tight_layout()
plt.savefig('clustering_evaluation.png', dpi=300, bbox_inches='tight')
plt.show()

# Select optimal K (you can adjust based on the plots)
optimal_k = 4  # Based on silhouette score and business interpretation
print(f"\n✓ Optimal number of clusters selected: {optimal_k}")

# ============================================================================
# SECTION 7: FINAL CLUSTERING MODELS
# ============================================================================

print("\n[STEP 8] TRAINING CLUSTERING MODELS")
print("-"*80)

# 1. K-Means Clustering
kmeans_final = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
kmeans_labels = kmeans_final.fit_predict(X_scaled)
clustering_features['KMeans_Cluster'] = kmeans_labels

print(f"✓ K-Means clustering completed")
print(f"  Silhouette Score: {silhouette_score(X_scaled, kmeans_labels):.3f}")

# 2. Hierarchical Clustering
hierarchical = AgglomerativeClustering(n_clusters=optimal_k)
hierarchical_labels = hierarchical.fit_predict(X_scaled)
clustering_features['Hierarchical_Cluster'] = hierarchical_labels

print(f"✓ Hierarchical clustering completed")
print(f"  Silhouette Score: {silhouette_score(X_scaled, hierarchical_labels):.3f}")

# 3. DBSCAN (for comparison)
dbscan = DBSCAN(eps=2, min_samples=5)
dbscan_labels = dbscan.fit_predict(X_scaled)
clustering_features['DBSCAN_Cluster'] = dbscan_labels

n_clusters_dbscan = len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0)
print(f"✓ DBSCAN clustering completed")
print(f"  Number of clusters: {n_clusters_dbscan}")
print(f"  Noise points: {list(dbscan_labels).count(-1)}")

# Use K-Means as final model
final_labels = kmeans_labels
clustering_features['FinalCluster'] = final_labels

# Save clustering model
joblib.dump(kmeans_final, 'kmeans_model.pkl')
print("✓ K-Means model saved")

# ============================================================================
# SECTION 8: CLUSTER ANALYSIS AND PROFILING
# ============================================================================

print("\n[STEP 9] CLUSTER ANALYSIS")
print("-"*80)

# Analyze each cluster
for cluster in range(optimal_k):
    cluster_data = clustering_features[clustering_features['FinalCluster'] == cluster]
    print(f"\n[CLUSTER {cluster}] - {len(cluster_data)} customers ({len(cluster_data)/len(clustering_features)*100:.1f}%)")
    print("-" * 60)
    
    print(f"  Recency (days): {cluster_data['Recency'].mean():.1f}")
    print(f"  Frequency (transactions): {cluster_data['Frequency'].mean():.1f}")
    print(f"  Monetary Value (£): {cluster_data['MonetaryTotal'].mean():.2f}")
    print(f"  Monthly Purchase (£): {cluster_data['MonthlyPurchaseAmount'].mean():.2f}")
    print(f"  Loyal Customers: {cluster_data['IsLoyal'].sum()} ({cluster_data['IsLoyal'].mean()*100:.1f}%)")
    print(f"  Frequent Buyers: {cluster_data['IsFrequentBuyer'].sum()} ({cluster_data['IsFrequentBuyer'].mean()*100:.1f}%)")
    print(f"  Top Category: {cluster_data['TopCategory'].mode()[0] if len(cluster_data['TopCategory'].mode()) > 0 else 'N/A'}")
    print(f"  Top Country: {cluster_data['Country'].mode()[0] if len(cluster_data['Country'].mode()) > 0 else 'N/A'}")

# Visualize clusters
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(12, 8))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=final_labels, 
                     cmap='viridis', alpha=0.6, s=50, edgecolors='black')
plt.colorbar(scatter, label='Cluster')
plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%} variance)')
plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%} variance)')
plt.title('Customer Segments (PCA Visualization)', fontsize=16, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.savefig('cluster_visualization.png', dpi=300, bbox_inches='tight')
plt.show()

# Cluster profiling heatmap
cluster_profile = clustering_features.groupby('FinalCluster')[
    ['Recency', 'Frequency', 'MonetaryTotal', 'MonthlyPurchaseAmount', 
     'IsLoyal', 'IsFrequentBuyer']].mean()

plt.figure(figsize=(10, 6))
sns.heatmap(cluster_profile.T, annot=True, fmt='.2f', cmap='YlOrRd', 
            cbar_kws={'label': 'Value'})
plt.title('Cluster Profiling Heatmap', fontsize=16, fontweight='bold')
plt.xlabel('Cluster')
plt.ylabel('Features')
plt.tight_layout()
plt.savefig('cluster_profile_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()

# Save clustered data
clustering_features.to_csv('customer_segments.csv', index=False)
print("\n✓ Customer segments saved to 'customer_segments.csv'")

# ============================================================================
# SECTION 9: CLUSTER PREDICTION MODEL
# ============================================================================

print("\n[STEP 10] BUILDING CLUSTER PREDICTION MODEL")
print("-"*80)

# Prepare data for classification
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, final_labels, test_size=0.2, random_state=42)

# Train Random Forest Classifier
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42, max_depth=10)
rf_classifier.fit(X_train, y_train)

# Predictions
y_pred = rf_classifier.predict(X_test)

# Evaluation
accuracy = accuracy_score(y_test, y_pred)
print(f"✓ Random Forest Classifier trained")
print(f"  Accuracy: {accuracy:.4f}")

print("\n[Classification Report]")
print(classification_report(y_test, y_pred))

# Confusion Matrix
plt.figure(figsize=(8, 6))
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False)
plt.title('Confusion Matrix', fontsize=16, fontweight='bold')
plt.xlabel('Predicted Cluster')
plt.ylabel('True Cluster')
plt.savefig('confusion_matrix.png', dpi=300, bbox_inches='tight')
plt.show()

# Feature Importance
feature_importance = pd.DataFrame({
    'Feature': feature_cols,
    'Importance': rf_classifier.feature_importances_
}).sort_values('Importance', ascending=False)

plt.figure(figsize=(10, 8))
plt.barh(range(len(feature_importance.head(15))), 
         feature_importance.head(15)['Importance'].values)
plt.yticks(range(len(feature_importance.head(15))), 
           feature_importance.head(15)['Feature'].values)
plt.xlabel('Importance')
plt.title('Top 15 Feature Importance for Cluster Prediction', 
          fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('feature_importance.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n[Top 10 Important Features]")
print(feature_importance.head(10))

# Save classifier
joblib.dump(rf_classifier, 'cluster_classifier.pkl')
print("✓ Cluster prediction model saved")

# Cross-validation
cv_scores = cross_val_score(rf_classifier, X_scaled, final_labels, cv=5)
print(f"\n✓ Cross-Validation Scores: {cv_scores}")
print(f"✓ Mean CV Score: {cv_scores.mean():.4f} (+/- {cv_scores.std() * 2:.4f})")

# ============================================================================
# SECTION 10: BUSINESS RECOMMENDATIONS
# ============================================================================

print("\n" + "="*80)
print("BUSINESS INSIGHTS AND RECOMMENDATIONS")
print("="*80)

# Cluster naming and insights
cluster_insights = {}

for cluster in range(optimal_k):
    cluster_data = clustering_features[clustering_features['FinalCluster'] == cluster]
    
    avg_recency = cluster_data['Recency'].mean()
    avg_frequency = cluster_data['Frequency'].mean()
    avg_monetary = cluster_data['MonetaryTotal'].mean()
    loyal_pct = cluster_data['IsLoyal'].mean() * 100
    
    # Cluster naming logic
    if avg_monetary > clustering_features['MonetaryTotal'].quantile(0.75) and avg_frequency > clustering_features['Frequency'].quantile(0.75):
        name = "Champions"
        recommendation = "Reward them. They are your best customers!"
    elif avg_recency < clustering_features['Recency'].quantile(0.25) and avg_frequency > clustering_features['Frequency'].median():
        name = "Loyal Customers"
        recommendation = "Upsell higher value products. Keep them engaged."
    elif avg_recency < clustering_features['Recency'].median() and avg_monetary > clustering_features['MonetaryTotal'].median():
        name = "Potential Loyalists"
        recommendation = "Offer membership/loyalty program. Increase frequency."
    else:
        name = "At Risk/Needs Attention"
        recommendation = "Re-engagement campaigns. Special offers to win them back."
    
    cluster_insights[cluster] = {
        'Name': name,
        'Size': len(cluster_data),
        'Percentage': f"{len(cluster_data)/len(clustering_features)*100:.1f}%",
        'Avg_Recency': f"{avg_recency:.0f} days",
        'Avg_Frequency': f"{avg_frequency:.1f} transactions",
        'Avg_Monetary': f"£{avg_monetary:.2f}",
        'Loyal_Percentage': f"{loyal_pct:.1f}%",
        'Top_Category': cluster_data['TopCategory'].mode()[0] if len(cluster_data['TopCategory'].mode()) > 0 else 'N/A',
        'Recommendation': recommendation
    }

# Display insights
for cluster, info in cluster_insights.items():
    print(f"\n{'='*80}")
    print(f"CLUSTER {cluster}: {info['Name']}")
    print(f"{'='*80}")
    print(f"  Size: {info['Size']} customers ({info['Percentage']})")
    print(f"  Average Recency: {info['Avg_Recency']}")
    print(f"  Average Frequency: {info['Avg_Frequency']}")
    print(f"  Average Monetary Value: {info['Avg_Monetary']}")
    print(f"  Loyal Customers: {info['Loyal_Percentage']}")
    print(f"  Favorite Category: {info['Top_Category']}")
    print(f"\n  💡 RECOMMENDATION: {info['Recommendation']}")

# Save insights to CSV
insights_df = pd.DataFrame(cluster_insights).T
insights_df.to_csv('cluster_insights.csv')
print("\n✓ Cluster insights saved to 'cluster_insights.csv'")

# ============================================================================
# SECTION 11: MODEL PERFORMANCE SUMMARY
# ============================================================================

print("\n" + "="*80)
print("MODEL PERFORMANCE SUMMARY")
print("="*80)

summary = {
    'Total Customers': len(clustering_features),
    'Features Used': len(feature_cols),
    'Clustering Algorithm': 'K-Means',
    'Number of Clusters': optimal_k,
    'Silhouette Score': f"{silhouette_score(X_scaled, final_labels):.4f}",
    'Davies-Bouldin Score': f"{davies_bouldin_score(X_scaled, final_labels):.4f}",
    'Calinski-Harabasz Score': f"{calinski_harabasz_score(X_scaled, final_labels):.2f}",
    'Prediction Model': 'Random Forest',
    'Prediction Accuracy': f"{accuracy:.4f}",
    'Cross-Validation Mean': f"{cv_scores.mean():.4f}"
}

for key, value in summary.items():
    print(f"  {key}: {value}")

# Save model summary
with open('model_summary.txt', 'w') as f:
    f.write("E-COMMERCE CUSTOMER SEGMENTATION MODEL SUMMARY\n")
    f.write("=" * 80 + "\n\n")
    for key, value in summary.items():
        f.write(f"{key}: {value}\n")
    f.write("\n" + "=" * 80 + "\n")
    f.write("CLUSTER INSIGHTS\n")
    f.write("=" * 80 + "\n\n")
    for cluster, info in cluster_insights.items():
        f.write(f"\nCluster {cluster}: {info['Name']}\n")
        f.write(f"  Size: {info['Size']} ({info['Percentage']})\n")
        f.write(f"  Recommendation: {info['Recommendation']}\n")

print("\n✓ Model summary saved to 'model_summary.txt'")

# ============================================================================
# SECTION 12: SAVE ALL ARTIFACTS
# ============================================================================

print("\n" + "="*80)
print("SAVING ALL ARTIFACTS")
print("="*80)

# Save feature column names
with open('feature_columns.txt', 'w') as f:
    for col in feature_cols:
        f.write(f"{col}\n")

print("✓ All models and artifacts saved:")
print("  - kmeans_model.pkl")
print("  - cluster_classifier.pkl")
print("  - standard_scaler.pkl")
print("  - label_encoder_top_category.pkl")
print("  - label_encoder_bottom_category.pkl")
print("  - label_encoder_country.pkl")
print("  - customer_features.csv")
print("  - customer_segments.csv")
print("  - cluster_insights.csv")
print("  - model_summary.txt")
print("  - feature_columns.txt")
print("  - All visualization plots")

print("\n" + "="*80)
print("PROJECT COMPLETED SUCCESSFULLY!")
print("="*80)

# ============================================================================
# SECTION 13: DEMONSTRATION - PREDICTING NEW CUSTOMER CLUSTER
# ============================================================================

print("\n" + "="*80)
print("DEMONSTRATION: PREDICTING CLUSTER FOR NEW CUSTOMERS")
print("="*80)

def predict_customer_cluster(customer_data):
    """
    Predict cluster for a new customer
    
    Parameters:
    customer_data: dict with customer features
    
    Returns:
    Predicted cluster and cluster name
    """
    # Load models
    kmeans = joblib.load('kmeans_model.pkl')
    rf_clf = joblib.load('cluster_classifier.pkl')
    scaler = joblib.load('standard_scaler.pkl')
    le_top = joblib.load('label_encoder_top_category.pkl')
    le_bottom = joblib.load('label_encoder_bottom_category.pkl')
    le_country = joblib.load('label_encoder_country.pkl')
    
    # Prepare features
    features = []
    for col in feature_cols:
        if col == 'TopCategory_Encoded':
            val = le_top.transform([customer_data.get('TopCategory', 'Unknown')])[0]
        elif col == 'BottomCategory_Encoded':
            val = le_bottom.transform([customer_data.get('BottomCategory', 'Unknown')])[0]
        elif col == 'Country_Encoded':
            val = le_country.transform([customer_data.get('Country', 'Unknown')])[0]
        else:
            val = customer_data.get(col, 0)
        features.append(val)
    
    # Scale features
    features_array = np.array(features).reshape(1, -1)
    features_scaled = scaler.transform(features_array)
    
    # Predict
    cluster = rf_clf.predict(features_scaled)[0]
    cluster_name = cluster_insights[cluster]['Name']
    recommendation = cluster_insights[cluster]['Recommendation']
    
    return cluster, cluster_name, recommendation

# Example prediction
print("\n[Example: New Customer Prediction]")
new_customer = {
    'Recency': 30,
    'Frequency': 15,
    'MonetaryTotal': 2500,
    'MonetaryAvg': 166.67,
    'MonetaryMin': 10,
    'MonetaryMax': 500,
    'MonthlyPurchaseAmount': 416.67,
    'TopCategorySpend': 1200,
    'TotalQuantity': 150,
    'PurchaseFrequencyPerMonth': 2.5,
    'CustomerAge': 180,
    'LifetimeValue': 3000,
    'LifetimeFrequency': 20,
    'IsLoyal': 1,
    'IsFrequentBuyer': 1,
    'TopCategory': 'Kitchen & Dining',
    'BottomCategory': 'Toys & Games',
    'Country': 'United Kingdom'
}

cluster_pred, cluster_name, recommendation = predict_customer_cluster(new_customer)
print(f"\nCustomer Profile:")
for key, value in new_customer.items():
    print(f"  {key}: {value}")

print(f"\n✓ Predicted Cluster: {cluster_pred}")
print(f"✓ Cluster Name: {cluster_name}")
print(f"✓ Recommendation: {recommendation}")

print("\n" + "="*80)
print("END OF ANALYSIS")
print("="*80)
```
